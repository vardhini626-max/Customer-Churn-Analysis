```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             roc_auc_score, classification_report, ConfusionMatrixDisplay,
                             RocCurveDisplay)

RANDOM_STATE = 42
os.makedirs("data", exist_ok=True)
os.makedirs("reports/figures", exist_ok=True)
plt.rcParams["figure.dpi"] = 100
```

## 1. Load / generate the data

If `data/telecom_churn.csv` already exists it is loaded; otherwise a synthetic dataset of 7,000 customers is generated and saved.

Columns: tenure, monthly/total charges, data usage, support calls, late payments, contract, internet service, payment method, tech support, paperless billing, senior citizen flag, and the target **`churn`** (1 = cancelled, 0 = stayed).

```python
def generate_data(n_rows=7000, seed=42):
    rng = np.random.default_rng(seed)

    contract = rng.choice(["Month-to-month", "One year", "Two year"], n_rows, p=[0.55, 0.25, 0.20])
    internet = rng.choice(["Fiber", "DSL", "None"], n_rows, p=[0.45, 0.35, 0.20])
    payment = rng.choice(["Electronic check", "Credit card", "Bank transfer", "Mailed check"],
                         n_rows, p=[0.35, 0.25, 0.25, 0.15])
    tech_support = rng.choice(["Yes", "No"], n_rows, p=[0.35, 0.65])
    paperless = rng.choice(["Yes", "No"], n_rows, p=[0.6, 0.4])
    senior = rng.choice([0, 1], n_rows, p=[0.84, 0.16])

    tenure = np.clip(rng.gamma(2.2, 14, n_rows).astype(int), 1, 72)
    base = np.select([internet == "Fiber", internet == "DSL", internet == "None"], [75, 52, 22])
    monthly = np.round(base + rng.normal(0, 8, n_rows) + 6 * (tech_support == "Yes"), 2)
    monthly = np.clip(monthly, 18, 125)
    total = np.round(monthly * tenure * rng.uniform(0.95, 1.05, n_rows), 2)
    data_gb = np.round(np.clip(rng.gamma(2.0, 12, n_rows) * (internet != "None"), 0, 200), 1)
    support_calls = rng.poisson(1.2 + 0.9 * (internet == "Fiber"), n_rows)
    late_payments = rng.poisson(0.4 + 0.5 * (payment == "Electronic check"), n_rows)

    # hidden churn mechanism (log-odds) plus noise
    z = (-1.1
         + 1.25 * (contract == "Month-to-month") - 1.1 * (contract == "Two year")
         + 0.55 * (internet == "Fiber") - 0.9 * (internet == "None")
         + 0.45 * (payment == "Electronic check") - 0.5 * (tech_support == "Yes")
         + 0.25 * senior + 0.012 * (monthly - 65) - 0.035 * tenure
         + 0.28 * support_calls + 0.35 * late_payments - 0.004 * data_gb
         + rng.normal(0, 0.6, n_rows))
    churn = (rng.random(n_rows) < 1 / (1 + np.exp(-z))).astype(int)

    df = pd.DataFrame({
        "customer_id": [f"CUST-{i:05d}" for i in range(1, n_rows + 1)],
        "tenure_months": tenure, "monthly_charges": monthly, "total_charges": total,
        "data_usage_gb": data_gb, "support_calls": support_calls, "late_payments": late_payments,
        "senior_citizen": senior, "contract": contract, "internet_service": internet,
        "payment_method": payment, "tech_support": tech_support, "paperless_billing": paperless,
        "churn": churn,
    })
    # a few missing values to practise imputation
    miss = rng.choice(n_rows, size=int(0.01 * n_rows), replace=False)
    df.loc[miss, "total_charges"] = np.nan
    return df


DATA_PATH = "data/telecom_churn.csv"
if not os.path.exists(DATA_PATH):
    generate_data().to_csv(DATA_PATH, index=False)

df = pd.read_csv(DATA_PATH)
df.head()
```

## 3. Preprocessing and train/test split

- **Numeric features:** fill missing values with the median, then standardise (important for Logistic Regression).
- **Categorical features:** fill missing values with the most frequent value, then one-hot encode.
- **Split:** 80% train / 20% test, *stratified* so both sets keep the same churn rate.

Everything is wrapped in a `Pipeline`, so the preprocessing is learned from the training data only (no data leakage).

```python
TARGET = "churn"
NUMERIC = ["tenure_months", "monthly_charges", "total_charges", "data_usage_gb",
           "support_calls", "late_payments", "senior_citizen"]
CATEGORICAL = ["contract", "internet_service", "payment_method", "tech_support", "paperless_billing"]

X = df[NUMERIC + CATEGORICAL]
y = df[TARGET]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=RANDOM_STATE)

def build_preprocessor():
    numeric_pipe = Pipeline([("impute", SimpleImputer(strategy="median")),
                             ("scale", StandardScaler())])
    categorical_pipe = Pipeline([("impute", SimpleImputer(strategy="most_frequent")),
                                 ("onehot", OneHotEncoder(handle_unknown="ignore"))])
    return ColumnTransformer([("num", numeric_pipe, NUMERIC),
                              ("cat", categorical_pipe, CATEGORICAL)])

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
print("Train:", X_train.shape, "| Test:", X_test.shape)
print(f"Churn rate  train: {y_train.mean():.1%}  test: {y_test.mean():.1%}")
```

## 4. Logistic Regression

Logistic Regression models the **log-odds** of churn as a linear combination of the features. `class_weight="balanced"` gives the minority class (churners) more weight so the model does not simply predict "stayed" for everyone. The regularisation strength `C` is tuned with 5-fold cross-validation using ROC-AUC.

```python
log_reg = Pipeline([
    ("prep", build_preprocessor()),
    ("clf", LogisticRegression(max_iter=1000, class_weight="balanced", random_state=RANDOM_STATE)),
])

lr_search = GridSearchCV(log_reg, {"clf__C": [0.01, 0.1, 1, 10]}, cv=cv, scoring="roc_auc", n_jobs=-1)
lr_search.fit(X_train, y_train)
best_lr = lr_search.best_estimator_

print("Best parameters:", lr_search.best_params_)
print(f"Cross-validated ROC-AUC: {lr_search.best_score_:.4f}")
```

## 5. Decision Tree

A Decision Tree learns a set of if/else rules. Unrestricted trees overfit, so we tune `max_depth` and `min_samples_leaf` (the minimum customers per leaf).

```python
tree = Pipeline([
    ("prep", build_preprocessor()),
    ("clf", DecisionTreeClassifier(class_weight="balanced", random_state=RANDOM_STATE)),
])

tree_search = GridSearchCV(
    tree,
    {"clf__max_depth": [3, 4, 5, 6, 8], "clf__min_samples_leaf": [10, 30, 60]},
    cv=cv, scoring="roc_auc", n_jobs=-1)
tree_search.fit(X_train, y_train)
best_tree = tree_search.best_estimator_

print("Best parameters:", tree_search.best_params_)
print(f"Cross-validated ROC-AUC: {tree_search.best_score_:.4f}")
```

## 6. Evaluation with classification metrics

For churn, "positive" = the customer **churned**.

| Metric | Question it answers | Formula |
|---|---|---|
| **Accuracy** | How often is the model right overall? | (TP + TN) / all |
| **Precision** | Of customers we *flagged* as churners, how many really churned? | TP / (TP + FP) |
| **Recall** | Of customers who *really churned*, how many did we catch? | TP / (TP + FN) |
| **F1** | Balance of precision and recall | 2PR / (P + R) |
| **ROC-AUC** | How well does the model rank churners above non-churners? | area under ROC curve |

Recall is usually the priority: missing a real churner (false negative) costs more than giving a retention offer to someone who would have stayed (false positive).

```python
def get_metrics(model):
    pred = model.predict(X_test)
    proba = model.predict_proba(X_test)[:, 1]
    return {
        "Accuracy": accuracy_score(y_test, pred),
        "Precision": precision_score(y_test, pred),
        "Recall": recall_score(y_test, pred),
        "F1": f1_score(y_test, pred),
        "ROC-AUC": roc_auc_score(y_test, proba),
    }

results = pd.DataFrame({"Logistic Regression": get_metrics(best_lr),
                        "Decision Tree": get_metrics(best_tree)}).T.round(4)
results.to_csv("reports/model_comparison.csv")
results
```

```python
for name, model in [("Logistic Regression", best_lr), ("Decision Tree", best_tree)]:
    print(f"===== {name} =====")
    print(classification_report(y_test, model.predict(X_test), target_names=["Stayed", "Churned"]))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
for ax, (name, model) in zip(axes, [("Logistic Regression", best_lr), ("Decision Tree", best_tree)]):
    ConfusionMatrixDisplay.from_predictions(y_test, model.predict(X_test),
                                            display_labels=["Stayed", "Churned"],
                                            cmap="Blues", ax=ax, colorbar=False)
    ax.set_title(name)
plt.tight_layout()
plt.savefig("reports/figures/confusion_matrices.png", dpi=150)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(6, 5))
RocCurveDisplay.from_estimator(best_lr, X_test, y_test, name="Logistic Regression", ax=ax)
RocCurveDisplay.from_estimator(best_tree, X_test, y_test, name="Decision Tree", ax=ax)
ax.plot([0, 1], [0, 1], "k--", alpha=0.4, label="Random guess")
ax.set_title("ROC curves (test set)")
ax.legend()
plt.tight_layout()
plt.savefig("reports/figures/roc_curves.png", dpi=150)
plt.show()
```

## 7. Interpreting the models

**Logistic Regression coefficients** (on standardised features): positive = increases churn risk, negative = reduces it.

```python
feature_names = [n.split("__", 1)[1] for n in best_lr.named_steps["prep"].get_feature_names_out()]

coefs = pd.Series(best_lr.named_steps["clf"].coef_[0], index=feature_names)
top = coefs.reindex(coefs.abs().sort_values(ascending=False).index[:12]).sort_values()

fig, ax = plt.subplots(figsize=(7, 5))
top.plot.barh(color=["#d9534f" if v > 0 else "#5cb85c" for v in top], ax=ax)
ax.set_title("Logistic Regression: top coefficients\n(red = more likely to churn)")
plt.tight_layout()
plt.savefig("reports/figures/lr_coefficients.png", dpi=150)
plt.show()
```

**Decision Tree feature importance and the first levels of the tree:**

```python
importances = pd.Series(best_tree.named_steps["clf"].feature_importances_,
                        index=feature_names).sort_values()

fig, ax = plt.subplots(figsize=(7, 5))
importances.tail(12).plot.barh(color="#4a90d9", ax=ax)
ax.set_title("Decision Tree: feature importance")
plt.tight_layout()
plt.savefig("reports/figures/tree_feature_importance.png", dpi=150)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(18, 8))
plot_tree(best_tree.named_steps["clf"], feature_names=feature_names,
          class_names=["Stayed", "Churned"], filled=True, rounded=True,
          max_depth=3, fontsize=8, ax=ax)
plt.tight_layout()
plt.savefig("reports/figures/decision_tree.png", dpi=130)
plt.show()
```

## 8. Adjusting the decision threshold

By default a customer is flagged as a churner when the predicted probability is at least 0.5. Changing this threshold trades precision against recall. A retention team with a limited budget may prefer higher precision; one that wants to catch as many churners as possible may prefer higher recall.

```python
proba = best_lr.predict_proba(X_test)[:, 1]
rows = []
for t in [0.3, 0.4, 0.5, 0.6, 0.7]:
    pred = (proba >= t).astype(int)
    rows.append({"Threshold": t,
                 "Precision": precision_score(y_test, pred),
                 "Recall": recall_score(y_test, pred),
                 "F1": f1_score(y_test, pred),
                 "Customers flagged": int(pred.sum())})
pd.DataFrame(rows).round(3)
```

## 9. Predicting a new customer

The fitted pipelines handle all preprocessing, so we can pass raw customer details directly.

```python
new_customer = pd.DataFrame([{
    "tenure_months": 3, "monthly_charges": 89.5, "total_charges": 268.5,
    "data_usage_gb": 14.0, "support_calls": 4, "late_payments": 2, "senior_citizen": 0,
    "contract": "Month-to-month", "internet_service": "Fiber",
    "payment_method": "Electronic check", "tech_support": "No", "paperless_billing": "Yes",
}])

prob = best_lr.predict_proba(new_customer)[0, 1]
print(f"Churn probability: {prob:.1%}")
print("Prediction:", "WILL LIKELY CHURN" if prob >= 0.5 else "likely to stay")
```

## 10. Conclusions

- **Logistic Regression** scored higher than the Decision Tree on every metric and is easy to explain through its coefficients.
- **Decision Tree** offers readable if/else rules but is a little less accurate here.
- **Accuracy alone is not enough** for churn. Precision, recall, F1 and ROC-AUC give a fuller picture, and both models use balanced class weights to favour recall.
- **Main churn drivers:** month-to-month contracts, short tenure, frequent support calls, late payments and electronic-check payment. Customers with long contracts and tech support are the most loyal.
- **Business use:** target retention offers (discounts, contract upgrades, proactive support) at customers with a high predicted churn probability.
